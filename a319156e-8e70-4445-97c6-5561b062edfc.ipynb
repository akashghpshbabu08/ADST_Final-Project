```python
total_rows = [
    "Total State (S)",
    "Total UT(S)",
    "Total All India"
]
data_clean = data[~data["State/UT"].isin(total_rows)].copy()
```

```python
data_clean.shape
```

```python
crime_columns = ["2020", "2021", "2022"]

for col in crime_columns:
    data_clean[col] = pd.to_numeric(
        data_clean[col],
        errors="coerce"
    )
```

```python
data_clean["Crime Change 2020-2022"] = (
    data_clean["2022"] - data_clean["2020"]
)
```

```python
data_clean["Crime Change % 2020-2022"] = (
    (data_clean["2022"] - data_clean["2020"])
    / data_clean["2020"]
) * 100
```

```python
data_clean["Crime Change % 2020-2022"] = (
    data_clean["Crime Change % 2020-2022"].round(2)
)
```

```python
data_clean = data_clean.rename(columns={
    "Sl. No.": "Serial_No",
    "State/UT": "State_UT",
    "Mid-Year Projected Population (in Lakhs) (2022)": "Population_Lakhs",
    "Rate of Cognizable Crimes (IPC) (2022)": "Crime_Rate_2022",
    "Chargesheeting Rate (2022)": "Chargesheeting_Rate_2022"
})
```

```python
columnName = '2020'
columnName = '2021'
columnName = '2022'

Q1 = data[columnName].quantile(0.25)
Q3 = data[columnName].quantile(0.75)

print("Q1:", Q1)
print("Q3:", Q3)

IQR = Q3 - Q1
print("IQR:", IQR)

LB = Q1 - 1.5 * IQR
print("Lower Bound:", LB)

UB = Q3 + 1.5 * IQR
print("Upper Bound:", UB)

outliers = data[
    (data[columnName] < LB) |
    (data[columnName] > UB)
]

print(f"No. of Outliers: {len(outliers)}")

display(outliers)
```
